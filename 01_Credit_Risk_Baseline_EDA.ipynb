# Define the Target Variable

```python
bad_statuses = [
    "Charged Off",
    "Default",
    "Late (31-120 days)",
    "Late (16-30 days)"
]

df_loan_data["target_default"] = df_loan_data["loan_status"].isin(bad_statuses).astype(int)
'''
0 Represents a "good" loan (not in the bad_statuses list).
1 Represents a "bad" loan (in the bad_statuses list).
'''
```

Check Class Imbalance

```python
df_loan_data["target_default"].value_counts(normalize=True)
```

Categorize by Year

```python
df_loan_data["issue_d"] = pd.to_datetime(df_loan_data["issue_d"], format="%b-%Y")
df_loan_data["issue_year"] = df_loan_data["issue_d"].dt.year
```

Check for coluns with high occurence of missing data

```python
missing = df_loan_data.isnull().mean().sort_values(ascending=False)
missing.head(20)
```

Flagging High Missing Features

```python
missing[missing > 0.3]
```

# Analyze Default Rates by Risk Drivers

## Group by Loan Grade and Probability of Default

```python
df_loan_data.groupby("grade")["target_default"].mean()
```

This grouping shows proportion of defaulted loans by grade, 0 being a good loan and 1 being a bad loan. The proportion indicates default rates

```python
sns.histplot(
    data=df_loan_data,
    x="dti",
    hue="target_default",
    bins=100,
    stat="density",
    common_norm=False
)
```

Density Plot indicates there is major overlap of "good" and "bad" loans within the same DTI bins. The data also shows that DTI is not relatively high for the loan data indicating invidiuals to not have a high Debt to Income ratio. DTI will not be a good indicator if a loan is like to default
