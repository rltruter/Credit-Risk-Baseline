```python
import pandas as pd
import os

directory_path = '/content/drive/MyDrive/Credit Risk Baseline Data'

# Define file paths for the second and third files
excel_file_path = os.path.join(directory_path, 'LCDataDictionary.xlsx')
csv_file_path = os.path.join(directory_path, 'loan.csv')

# Read the Excel file into a DataFrame
try:
    df_data_dictionary = pd.read_excel(excel_file_path)
    print(f"Successfully loaded '{os.path.basename(excel_file_path)}' into df_data_dictionary.")
    print("First 5 rows of df_data_dictionary:")
    display(df_data_dictionary.head())
except FileNotFoundError:
    print(f"Error: The file '{excel_file_path}' was not found.")
except Exception as e:
    print(f"Error loading Excel file: {e}")
```

```python
# Read the CSV file into a DataFrame
try:
    df_loan_data = pd.read_csv(csv_file_path,low_memory = False)
    print(f"\nSuccessfully loaded '{os.path.basename(csv_file_path)}' into df_loan_data.")
    print("First 5 rows of df_loan_data:")
    display(df_loan_data.head())
except FileNotFoundError:
    print(f"Error: The file '{csv_file_path}' was not found.")
except Exception as e:
    print(f"Error loading CSV file: {e}")
```

# Define the Target Variable

```python
bad_statuses = [
    "Charged Off",
    "Default",
    "Late (31-120 days)",
    "Late (16-30 days)"
]

df_loan_data["target_default"] = df_loan_data["loan_status"].isin(bad_statuses).astype(int)
'''
0 Represents a "good" loan (not in the bad_statuses list).
1 Represents a "bad" loan (in the bad_statuses list).
'''
```

Check Class Imbalance

```python
df_loan_data["target_default"].value_counts(normalize=True)
```

Categorize by Year

```python
df_loan_data["issue_d"] = pd.to_datetime(df_loan_data["issue_d"], format="%b-%Y")
df_loan_data["issue_year"] = df_loan_data["issue_d"].dt.year
```

Check for coluns with high occurence of missing data

```python
missing = df_loan_data.isnull().mean().sort_values(ascending=False)
missing.head(20)
```
